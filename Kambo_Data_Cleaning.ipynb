```python
import pandas as pd
import numpy as np

# ✅ Load dataset
df = long_df_merged.copy()

# ✅ Rename the dummy CEQ column to correct label
df.rename(columns={
    'LOOKING BACK ON THE ENTIRETY OF YOUR SESSIONS, PLEASE RATE THE DEGREE TO WHICH AT ANY TIME DURING THAT SESSION YOU EXPERIENCED THE FOLLOWING PHENOMENA. ANSWER EACH QUESTION ACCORDING TO YOUR FEELINGS, THOUGHTS, AND EXPERIENCES AT THE TIME OF THE SESSION. - CLICK TO WRITE STATEMENT 26 | INSTRUCTIONS_26':
    'LOOKING BACK... - I FELT FRIGHTENED | INSTRUCTIONS_26'
}, inplace=True)

# ✅ Label mapping
label_map = {
    'none at all': 0,
    'none; not at all': 0,
    'slight': 1,
    'so slight cannot decide': 1,
    'so slight, cannot detect': 1,
    'moderate': 2,
    'strong': 3,
    'strong (equivalent in degree to any other strong experience)': 3,
    'very strong': 4,
    'extreme': 5,
    'extreme (more than any other time in my life)': 5,
    'extreme (more than ever before in my life)': 5,
    'complete': 5,
    '': np.nan,
    'nan': np.nan,
    'NaN': np.nan
}

# ✅ Identify CEQ and MEQ columns
ceq_exact_phrases = [
    'isolation and loneliness', 'sadness', 'feeling my heart beating',
    'i had the feeling something horrible would happen', 'feeling my body shake',
    'feelings of grief', 'experience of fear', 'fear that i might lose my mind or go insane',
    'i felt like crying', 'feeling of isolation from people and things', 'feelings of despair',
    'i had the feeling that people were plotting against me', 'i was afraid that the state i was in would last forever',
    'anxiousness', 'i felt shaky inside', 'i had the profound experience of my own death',
    'i felt my heart beating irregularly or skipping beats', 'pressure or weight in my chest or abdomen',
    'i experienced a decreased sense of sanity', 'i felt as if i was dead or dying', 'panic',
    'experience of antagonism toward people around me', 'despair', 'i felt isolated from everything and everyone',
    'emotional and/or physical suffering', 'i felt frightened'
]

ceq_cols = [
    col for col in df.columns
    if 'looking back' in col.lower() and any(phrase in col.lower() for phrase in ceq_exact_phrases)
]

meq_cols = [
    col for col in df.columns
    if 'looking back' in col.lower() and col not in ceq_cols
]

print(f"✔️ CEQ columns: {len(ceq_cols)}")
print(f"✔️ MEQ columns: {len(meq_cols)}")

# ✅ Map responses to numbers
def map_responses(df, columns):
    for col in columns:
        df[col] = (
            df[col]
            .astype(str)
            .str.strip()
            .str.lower()
            .replace(label_map)
        )
    return df

df = map_responses(df, meq_cols + ceq_cols)

# ✅ Clean column names for publication
meq_rename_map = {
    col: 'MEQ30_' + col.split(' - ')[-1].split(' | ')[0].strip().replace(' ', '').replace('"', '').replace("'", "")
    for col in meq_cols
}

ceq_rename_map = {
    col: 'CEQ30_' + col.split(' - ')[-1].split(' | ')[0].strip().replace(' ', '').replace('"', '').replace("'", "")
    for col in ceq_cols
}

df = df.rename(columns={**meq_rename_map, **ceq_rename_map})

meq_cols_clean = list(meq_rename_map.values())
ceq_cols_clean = list(ceq_rename_map.values())

# ✅ MEQ Scoring
ineffability_items = [col for col in meq_cols_clean if 'describ' in col.lower() or 'difficulttocommunicate' in col.lower()]
mystical_items = [col for col in meq_cols_clean if any(kw in col.lower() for kw in [
    'oneness', 'ultimatereality', 'sacred', 'reverence', 'purebeing', 'spiritualheight',
    'unity', 'fusion', 'eternity', 'insight', 'presence', 'life', 'allisone'
])]
positive_items = [col for col in meq_cols_clean if any(kw in col.lower() for kw in [
    'amazement', 'tenderness', 'peace', 'ecstasy', 'awe', 'joy'
])]
transcendence_items = [col for col in meq_cols_clean if any(kw in col.lower() for kw in [
    'time', 'space', 'timelessness', 'realm', 'awarenessofwhereyouare'
])]

df['MEQ30_Ineffability'] = df[ineffability_items].astype(float).mean(axis=1)
df['MEQ30_MysticalExperience'] = df[mystical_items].astype(float).mean(axis=1)
df['MEQ30_PositiveMood'] = df[positive_items].astype(float).mean(axis=1)
df['MEQ30_TranscendenceOfTimeAndSpace'] = df[transcendence_items].astype(float).mean(axis=1)
df['MEQ30_TotalScore'] = df[meq_cols_clean].astype(float).mean(axis=1)

# ✅ CEQ Scoring
isolation_items = [col for col in ceq_cols_clean if 'isolation' in col.lower() or 'loneliness' in col.lower()]
fear_items = [col for col in ceq_cols_clean if 'fear' in col.lower() or 'panic' in col.lower() or 'frightened' in col.lower() or 'anxiousness' in col.lower()]
grief_items = [col for col in ceq_cols_clean if 'grief' in col.lower() or 'sadness' in col.lower() or 'crying' in col.lower() or 'despair' in col.lower()]
insanity_items = [col for col in ceq_cols_clean if 'insane' in col.lower() or 'sanity' in col.lower() or 'mind' in col.lower()]
death_items = [col for col in ceq_cols_clean if 'death' in col.lower() or 'dying' in col.lower()]
paranoia_items = [col for col in ceq_cols_clean if 'paranoia' in col.lower() or 'plott' in col.lower()]
physical_items = [col for col in ceq_cols_clean if any(kw in col.lower() for kw in [
    'body', 'physical', 'heartbeating', 'shaky', 'pressure', 'suffering', 'pain'
])]

# Helper function to score a subscale using the 30% missing rule
def score_subscale(df, item_list, subscale_name, min_prop_complete=0.7):
    item_array = df[item_list].astype(float)
    n_items = len(item_list)
    valid_count = item_array.notna().sum(axis=1)
    score = item_array.mean(axis=1)
    score = score.where(valid_count >= n_items * min_prop_complete)
    df[subscale_name] = score
    return df

# ✅ Score MEQ30 subscales
df = score_subscale(df, ineffability_items, 'MEQ30_Ineffability', min_prop_complete=0.7)
df = score_subscale(df, mystical_items, 'MEQ30_MysticalExperience', min_prop_complete=0.7)
df = score_subscale(df, positive_items, 'MEQ30_PositiveMood', min_prop_complete=0.7)
df = score_subscale(df, transcendence_items, 'MEQ30_TranscendenceOfTimeAndSpace', min_prop_complete=0.7)

# ✅ Compute MEQ30 Total only if all 4 subscales are present
df['MEQ30_TotalScore'] = df[[
    'MEQ30_Ineffability',
    'MEQ30_MysticalExperience',
    'MEQ30_PositiveMood',
    'MEQ30_TranscendenceOfTimeAndSpace'
]].mean(axis=1).where(df[[
    'MEQ30_Ineffability',
    'MEQ30_MysticalExperience',
    'MEQ30_PositiveMood',
    'MEQ30_TranscendenceOfTimeAndSpace'
]].notna().all(axis=1))

# ✅ Score CEQ30 subscales
df = score_subscale(df, isolation_items, 'CEQ30_Isolation_SUM', min_prop_complete=0.7)
df = score_subscale(df, fear_items, 'CEQ30_Fear_SUM', min_prop_complete=0.7)
df = score_subscale(df, grief_items, 'CEQ30_Grief_SUM', min_prop_complete=0.7)
df = score_subscale(df, insanity_items, 'CEQ30_Insanity_SUM', min_prop_complete=0.7)
df = score_subscale(df, death_items, 'CEQ30_Death_SUM', min_prop_complete=0.7)
df = score_subscale(df, paranoia_items, 'CEQ30_Paranoia_SUM', min_prop_complete=0.7)
df = score_subscale(df, physical_items, 'CEQ30_PhysicalDistress_SUM', min_prop_complete=0.7)

# ✅ Compute CEQ30 Total only if all 7 subscales are present
df['CEQ30_Total_SUM'] = df[[
    'CEQ30_Isolation_SUM', 'CEQ30_Fear_SUM', 'CEQ30_Grief_SUM',
    'CEQ30_Insanity_SUM', 'CEQ30_Death_SUM', 'CEQ30_Paranoia_SUM',
    'CEQ30_PhysicalDistress_SUM'
]].mean(axis=1).where(df[[
    'CEQ30_Isolation_SUM', 'CEQ30_Fear_SUM', 'CEQ30_Grief_SUM',
    'CEQ30_Insanity_SUM', 'CEQ30_Death_SUM', 'CEQ30_Paranoia_SUM',
    'CEQ30_PhysicalDistress_SUM'
]].notna().all(axis=1))

# ✅ Drop item-level columns after scoring
df.drop(columns=meq_cols_clean + ceq_cols_clean, inplace=True)

# ✅ Save or return updated DataFrame
df.to_csv("Kambo_df_with_MEQ_and_CEQ_scored_thresholded.csv", index=False)
```

```python
import pandas as pd
import numpy as np

### 📥 Load Data
df_columns = pd.read_csv('/content/drive/MyDrive/Kambo/ColumnNames.csv')

### 🧼 Clean & Prepare Column Names

def safe_str(x):
    return str(x) if pd.notna(x) else ''

def clean_column(col):
    col = safe_str(col)
    return (
        col.lower()
        .split('|')[0]
        .replace('click to write the question text - ', '')
        .replace('with respect to your previous kambo session, please drag the slider to indicate how much you agree with the following statements. - ', '')
        .strip()
        .replace('.', '')
        .replace(' ', '_')
    )

# Extract mappings from the columns file
columns_100_point = list(df_columns.columns)
columns_5_point = list(df_columns.iloc[0])

hundred_cols = [col for col in columns_100_point if pd.notna(col)]
five_cols = [col for col in columns_5_point if pd.notna(col)]

# Create rename maps
hundred_map = {col: clean_column(col) + '_100scale' for col in hundred_cols if col in df.columns}
five_map = {col: clean_column(col) + '_5scale' for col in five_cols if col in df.columns}

### 🔁 Clean Values for 5-point Columns

def clean_5scale_value(x):
    x = str(x).strip().lower()
    replacements = {
        'yes, much more than usualy': 'yes, much more than usually',
        'yes, much more than usually': 100,
        'no, not more than usually': 0,
        '1': 0, '2': 25, '3': 50, '4': 75, '5': 100,
    }
    return replacements.get(x, np.nan)

for col in five_cols:
    if col in df.columns:
        df[col] = df[col].apply(clean_5scale_value)

### 🏷️ Rename Columns
df = df.rename(columns={**hundred_map, **five_map})

# Drop duplicate columns
df = df.loc[:, ~df.columns.duplicated()]

### 🔀 Combine 5-point and 100-point into unified items
dasc_items = list(set(
    [v.replace('_100scale', '') for v in hundred_map.values()] +
    [v.replace('_5scale', '') for v in five_map.values()]
))

for item in dasc_items:
    col_100 = item + '_100scale'
    col_5 = item + '_5scale'
    if col_100 in df.columns and col_5 in df.columns:
        df[item] = df[col_100].combine_first(df[col_5])
    elif col_100 in df.columns:
        df[item] = df[col_100]
    elif col_5 in df.columns:
        df[item] = df[col_5]

df[dasc_items] = df[dasc_items].apply(pd.to_numeric, errors='coerce')

### 🧠 Define 5DASC Subscales
subscales = {
    "5DASC_OceanicBoundlessness": [
        "i_experienced_a_touch_of_eternity",
        "everything_seemed_to_unify_into_a_oneness",
        "conflicts_and_contradictions_seemed_to_dissolve",
        "i_felt_one_with_my_surroundings",
        "i_experienced_an_all-embracing_love",
        "worries_and_anxieties_of_everyday_life_felt_unimportant",
        "i_felt_extraordinary_powers_within_myself",
        "my_experience_had_religious_aspects_to_it",
        "i_felt_totally_free_and_released_from_all_obligations",
        "i_experienced_profound_inner_peace"
    ],
    "5DASC_AnxiousEgoDissolution": [
        "i_felt_tormented",
        "i_felt_threatened",
        "i_felt_as_if_dark_forces_had_overtaken_me",
        "i_was_afraid_that_the_state_i_was_in_would_last_forever",
        "i_was_scared_without_knowing_exactly_why",
        "i_felt_as_if_i_were_paralyzed",
        "i_felt_that_i_was_on_the_verge_of_unconsciousness",
        "i_felt_like_a_puppet_or_marionette"
    ],
    "5DASC_VisionaryRestructuralization": [
        "i_saw_regular_patterns_with_closed_eyes_or_in_complete_darkness",
        "i_saw_colors_with_closed_eyes_or_in_complete_darkness",
        "i_saw_brightness_or_flashes_of_light_with_closed_eyes_or_in_complete_darkness",
        "i_could_see_images_from_my_memory_or_imagination_with_extreme_clarity",
        "everything_around_me_seemed_to_be_animated_with_life",
        "objects_in_my_surroundings_engaged_me_emotionally_much_more_than_usual"
    ],
    "5DASC_AuditoryAlterations": [
        "i_heard_single_words_without_knowing_where_they_came_from",
        "i_heard_complete_sentences_without_knowing_where_they_came_from",
        "i_heard_my_thoughts_as_if_i_had_spoken_them_out_loud",
        "i_heard_music_without_knowing_where_it_came_from",
        "i_heard_voices_that_did_not_come_from_the_surroundings_as_usual",
        "i_heard_something_like_a_buzzing,_swooshing,_or_humming_without_recognizing_the_cause"
    ],
    "5DASC_ReductionOfVigilance": [
        "i_felt_sleepy",
        "i_felt_drowsy",
        "i_felt_like_i_do_shortly_before_falling_asleep",
        "i_felt_i_was_about_to_fall_asleep",
        "i_felt_drunk",
        "i_felt_as_if_i_was_half-asleep",
        "my_thoughts_and_actions_were_slowed_down",
        "i_felt_exhausted"
    ]
}

### 🧮 Score Each Subscale with 70% Rule
def score_subscale(df, items, subscale_name, min_prop_complete=0.7):
    matched_items = [item for item in items if item in df.columns]
    n_items = len(matched_items)
    item_data = df[matched_items].astype(float)
    valid_counts = item_data.notna().sum(axis=1)
    score = item_data.mean(axis=1)
    score = score.where(valid_counts >= n_items * min_prop_complete)
    df[f'{subscale_name}_Score'] = score
    return df

for subscale_name, items in subscales.items():
    df = score_subscale(df, items, subscale_name, min_prop_complete=0.7)

subscale_score_cols = [f'{s}_Score' for s in subscales.keys()]
df["5DASC_TotalScore"] = df[subscale_score_cols].mean(axis=1).where(df[subscale_score_cols].notna().all(axis=1))

### 💾 Save Output
df_clean = df.drop(columns=[c for c in df.columns if c.endswith('_100scale') or c.endswith('_5scale') or c in dasc_items], errors='ignore')
df_clean.to_csv("Kambo_with_5DASC_scored_clean.csv", index=False)
```

```python
# Step 1: Clean MAAS column names
def clean_maas_column(col):
    return col.split('|')[-1].strip().lower()

# Identify and map MAAS columns
maas_cols_raw = [col for col in df_clean.columns if 'MAAS_' in col]
maas_cols_clean = [clean_maas_column(c) for c in maas_cols_raw]
maas_col_map = dict(zip(maas_cols_clean, maas_cols_raw))

# Step 2: Define the text-to-score mapping
maas_map = {
    'almost never': 1,
    'very infrequently': 2,
    'somewhat infrequently': 3,
    'somewhat frequently': 4,
    'very frequently': 5,
    'almost always': 6
}

# Step 3: Apply mapping to each MAAS column
for col in maas_cols_raw:
    df_clean[col] = df_clean[col].astype(str).str.strip().str.lower().map(maas_map)

# Step 4: Compute MAAS Total Score (30% missing rule = at least 11 of 15 valid)
def score_maas_total(df_clean, maas_cols_raw, output_col='MAAS_Total', min_valid=11):
    item_data = df_clean[maas_cols_raw].astype(float)
    valid_counts = item_data.notna().sum(axis=1)
    mean_score = item_data.mean(axis=1)
    df_clean[output_col] = mean_score.where(valid_counts >= min_valid)
    return df_clean

# Step 5: Score and export
df_maas = score_maas_total(df_clean, maas_cols_raw, output_col='MAAS_Total')
df_maas_clean = df_maas.drop(columns=maas_cols_raw, errors='ignore')
df_maas_clean.to_csv("Kambo_with_MAAS_scored.csv", index=False)
```

```python
df = df_maas_clean.copy()

# Step 1: Define response mapping and reverse-score item numbers
ryff_map = {
    'strongly disagree': 1,
    'somewhat disagree': 2,
    'a little disagree': 3,
    'neither agree nor disagree': 4,
    'a little agree': 5,
    'somewhat agree': 6,
    'strongly agree': 7
}

reverse_items = [
    5, 8, 9, 10, 11, 12, 14, 15, 16,
    18, 19, 20, 23, 24, 25, 26, 29, 32, 33
]

# Step 2: Identify Ryff columns and create reverse-score list
ryff_cols_raw = [col for col in df.columns if '| RYFF_' in col]
reverse_cols = [col for col in ryff_cols_raw if any(col.strip().endswith(f'RYFF_{i}') for i in reverse_items)]

# Step 3: Clean and map all Ryff columns to numeric values
for col in ryff_cols_raw:
    df[col] = df[col].astype(str).str.strip().str.lower().map(ryff_map)

# Step 4: Reverse-score appropriate items
for col in reverse_cols:
    df[col] = df[col].apply(lambda x: 8 - x if pd.notnull(x) else np.nan)

# Step 5: Score total Ryff using 30% missing rule (≥ 30 of 42 answered)
def score_ryff_total(df, cols, output_col='RYFF_Total', min_valid=30):
    data = df[cols].astype(float)
    valid_counts = data.notna().sum(axis=1)
    mean_score = data.mean(axis=1)
    df[output_col] = mean_score.where(valid_counts >= min_valid)
    return df

df = score_ryff_total(df, ryff_cols_raw, output_col='RYFF_Total')

# Step 6: Subscale definitions
ryff_subscales = {
    "Autonomy": [1, 7, 13, 19, 25, 31, 37],
    "EnvironmentalMastery": [2, 8, 14, 20, 26, 32, 38],
    "PersonalGrowth": [3, 9, 15, 21, 27, 33, 39],
    "PositiveRelations": [4, 10, 16, 22, 28, 34, 40],
    "PurposeInLife": [5, 11, 17, 23, 29, 35, 41],
    "SelfAcceptance": [6, 12, 18, 24, 30, 36, 42],
}

# Step 7: Create item map (e.g., 'RYFF_1' → actual column name)
ryff_col_map = {
    f"RYFF_{i}": col for i in range(1, 43)
    for col in df.columns if col.strip().endswith(f'RYFF_{i}')
}

# Step 8: Subscale scoring with 30% missing rule
def score_subscale(df, col_map, item_nums, name, min_valid_ratio=0.7):
    cols = [col_map[f"RYFF_{i}"] for i in item_nums if f"RYFF_{i}" in col_map]
    valid_counts = df[cols].notna().sum(axis=1)
    score = df[cols].mean(axis=1)
    score = score.where(valid_counts >= len(cols) * min_valid_ratio)
    df[f"RYFF_{name}_Score"] = score
    return df

for name, item_nums in ryff_subscales.items():
    df = score_subscale(df, ryff_col_map, item_nums, name)

# Final Step: Drop item-level columns and export
df_ryff = df.drop(columns=ryff_cols_raw, errors='ignore')
df_ryff.to_csv("Kambo_with_Ryff_scored.csv", index=False)
```

```python
import re
from collections import defaultdict

# Reload the session dataset
df_session = df_bpi_cleaned.copy()

# Define point and location columns
location_columns = ['location of points', 'location of points.1', 'location of points.2']
point_columns = ['# of points', '# of points.1', '# of points.2']

# Updated keyword-region mapping
region_keywords = {
    "head": [
        "ear", "ears", "auricular", "shen men", "point zero", "cerebral",
        "master oscillation", "thalamus", "forehead", "face"
    ],
    "neck": ["neck", "throat"],
    "thorax": [
        "chest", "sternum", "heart", "lung", "heart chakra",
        "back", "upper back", "center back", "mid spine", "upper spine"
    ],
    "abdomen": [
        "abdomen", "solar plexus", "mid back", "front of chest",
        "stomach", "solar"
    ],
    "pelvis": [
        "sacrum", "sacral", "pelvis", "root", "lower back", "low back",
        "root chakra", "lower spine"
    ],
    "upper_extremity": [
        "shoulder", "arm", "upper arm", "l arm", "r arm", "left arm",
        "right arm", "hand", "elbow", "wrist"
    ],
    "lower_extremity": [
        "leg", "ankle", "knee", "lower leg", "foot", "feet", "toe"
    ]
}

# Reverse mapping for keyword search
keyword_to_region = {}
for region, keywords in region_keywords.items():
    for kw in keywords:
        keyword_to_region[kw] = region

# Function to parse locations
def parse_location_string(text):
    point_map = defaultdict(int)
    text = str(text).lower().strip()
    matches = re.findall(r'(\d+)\s+(?:on\s+)?([\w\s\-]+)', text)
    for num_str, body_part in matches:
        num = int(num_str)
        for kw, region in keyword_to_region.items():
            if re.search(rf"\b{re.escape(kw)}\b", body_part):
                point_map[region] += num
                break
        else:
            point_map["unclassified"] += num
    return point_map

# List of regions
regions = list(region_keywords.keys())

# Initialize per-session region columns
for session in [1, 2, 3]:
    for region in regions:
        df_session[f'Session{session}_{region}'] = 0

# Apply region classification per session
for idx, row in df_session.iterrows():
    for session_idx, (loc_col, pts_col) in enumerate(zip(location_columns, point_columns), start=1):
        text = str(row.get(loc_col, '')).strip().lower()
        parsed = parse_location_string(text)
        for region in regions:
            df_session.at[idx, f'Session{session_idx}_{region}'] += parsed.get(region, 0)

# Drop original raw location and point columns
df_session_cleaned = df_session.drop(columns=location_columns + point_columns, errors='ignore')

# Export cleaned dataframe
df_session_cleaned.to_csv("Kambo_session_regions_scored.csv", index=False)
```

```python
df_demo = df_standardized.copy()

# ------------------------
# Helper: convert height to inches
# ------------------------
def parse_height(height):
    if pd.isna(height):
        return np.nan

    height = str(height).lower().strip().replace("’", "'").replace("”", '"')
    height = re.sub(r'[^a-z0-9\'\"\s.]', '', height)

    # Handle spelled-out cases
    if height in ['six', '6']:
        return 72
    if height in ['five', '5']:
        return 60

    # Handle common patterns
    match_ft_in = re.search(r'(\d)\s*ft[\s\.]*(\d{1,2})?', height)
    if match_ft_in:
        ft = int(match_ft_in.group(1))
        inches = int(match_ft_in.group(2)) if match_ft_in.group(2) else 0
        return ft * 12 + inches

    match_ft_sym = re.search(r"(\d)'\s*(\d{1,2})?", height)
    if match_ft_sym:
        ft = int(match_ft_sym.group(1))
        inches = int(match_ft_sym.group(2)) if match_ft_sym.group(2) else 0
        return ft * 12 + inches

    match_dec = re.match(r'(\d+)[.,](\d+)', height)
    if match_dec:
        try:
            return round(float(height) * 12)
        except:
            return np.nan

    return np.nan

# ------------------------
# Helper: convert weight to lbs
# ------------------------
def parse_weight(weight):
    if pd.isna(weight):
        return np.nan
    weight = str(weight).lower().replace('lbs', '').replace('lb', '').strip()
    try:
        val = float(re.findall(r'\d+', weight)[0])
        return val if 80 <= val <= 400 else np.nan
    except:
        return np.nan

# ------------------------
# Helper: scale mappings
# ------------------------
edu_map = {
    'high school': 1,
    'college diploma or technical/trade school': 2,
    'university (e.g., bachelor’s degree)': 3,
    'professional degree (e.g., md, pharmd, llb, jd)': 4,
    'post-graduate degree (e.g., master’s or phd)': 5
}

income_map = {
    'less than $24,999 annually': 1,
    '$25,000-$50,000 annually': 2,
    '$50,000-$75,000 annually': 3,
    '$75,000-$100,000 annually': 4,
    '$100,000-$125,000 annually': 5,
    '$125,000-$150,000 annually': 6,
    'more than $150,000 annually': 7
}

binary_map = {'yes': 1, 'no': 0}
cannabis_map = {'never': 0, 'rarely': 1, 'often': 2, 'daily': 3}
gender_map = {'cis-female': 0, 'cis-male': 1, 'other': 2}
sex_map = {'female': 0, 'male': 1}
nervous_map = {'not nervous at all': 0, 'extremely nervous': 8}
excited_map = {'extremely excited': 8}

# ------------------------
# Standardization Logic
# ------------------------
df_standardized['Age'] = pd.to_numeric(df_demo["WHAT IS YOUR AGE? | DEMO.1."], errors='coerce')
df_standardized['Sex'] = df_demo["WHAT BIOLOGICAL SEX WERE YOU ASSIGNED AT BIRTH? | DEMO.2."].str.lower().map(sex_map)
df_standardized['Gender'] = df_demo["WHAT GENDER DO YOU IDENTIFY WITH? - SELECTED CHOICE | DEMO.3."].str.lower().map(gender_map)
df_standardized['Education'] = df_demo["WHAT IS YOUR HIGHEST LEVEL OF EDUCATION COMPLETED? | DEMO.5."].str.lower().map(edu_map)
df_standardized['Income'] = df_demo["WHAT IS YOUR GROSS HOUSEHOLD INCOME? | DEMO.6."].str.lower().map(income_map)
df_standardized['Height_in'] = df_demo["WHAT IS YOUR APPROXIMATE HEIGHT (FT, IN)? | DEMO.7."].apply(parse_height)
df_standardized['Weight_lb'] = df_demo["WHAT IS YOUR APPROXIMATE WEIGHT (LBS)? | DEMO.8."].apply(parse_weight)
df_standardized['Lyme_DX'] = df_demo["HAVE YOU BEEN DIAGNOSED WITH LYME DISEASE BY A LICENSED HEALTHCARE PROFESSIONAL? | DEMO.10."].str.lower().map(binary_map)
df_standardized['Post_Lyme_DX'] = df_demo["HAVE YOU BEEN DIAGNOSED WITH POST-TREATMENT LYME SYNDROME BY A LICENSED HEALTHCARE PROFESSIONAL? | DEMO.11."].str.lower().map(binary_map)
df_standardized['Psychedelic_Use'] = df_demo["HAVE YOU EVER USED A PSYCHEDELIC SUBSTANCE SUCH AS LSD, PSILOCYBIN (MAGIC) MUSHROOMS, MESCALINE (PEYOTE), DMT, OR AYAHUASCA? | DEMO.13."].str.lower().map(binary_map)
df_standardized['Cannabis_Use'] = df_demo["DO YOU USE CANNABIS? | DEMO.14."].str.lower().map(cannabis_map)

# Nervousness and excitement
df_standardized['Nervous'] = df_demo["PLEASE SELECT THE OPTION THAT REFLECTS HOW NERVOUS YOU ARE FEELING ABOUT YOUR UPCOMING KAMBO EXPERIENCES. | DEMO.16."].str.lower().replace(nervous_map).apply(pd.to_numeric, errors='coerce')
df_standardized['Excited'] = df_demo["PLEASE SELECT THE OPTION THAT REFLECTS HOW EXCITED YOU ARE FEELING ABOUT YOUR UPCOMING KAMBO EXPERIENCES. | DEMO.17."].str.lower().replace(excited_map).apply(pd.to_numeric, errors='coerce')

# BMI
height_m = df_standardized['Height_in'] * 0.0254
weight_kg = df_standardized['Weight_lb'] * 0.453592
df_standardized['BMI'] = weight_kg / (height_m ** 2)

columns_to_drop = [
    "WHAT IS YOUR AGE? | DEMO.1.",
    "WHAT BIOLOGICAL SEX WERE YOU ASSIGNED AT BIRTH? | DEMO.2.",
    "WHAT GENDER DO YOU IDENTIFY WITH? - SELECTED CHOICE | DEMO.3.",
    "WHAT RACE DO YOU IDENTIFY AS? | DEMO.4.",
    "WHAT IS YOUR HIGHEST LEVEL OF EDUCATION COMPLETED? | DEMO.5.",
    "WHAT IS YOUR GROSS HOUSEHOLD INCOME? | DEMO.6.",
    "WHAT IS YOUR APPROXIMATE HEIGHT (FT, IN)? | DEMO.7.",
    "WHAT IS YOUR APPROXIMATE WEIGHT (LBS)? | DEMO.8.",
    "HAVE YOU BEEN DIAGNOSED WITH LYME DISEASE BY A LICENSED HEALTHCARE PROFESSIONAL? | DEMO.10.",
    "HAVE YOU BEEN DIAGNOSED WITH POST-TREATMENT LYME SYNDROME BY A LICENSED HEALTHCARE PROFESSIONAL? | DEMO.11.",
    "HAVE YOU EVER USED A PSYCHEDELIC SUBSTANCE SUCH AS LSD, PSILOCYBIN (MAGIC) MUSHROOMS, MESCALINE (PEYOTE), DMT, OR AYAHUASCA? | DEMO.13.",
    "DO YOU USE CANNABIS? | DEMO.14.",
    "PLEASE SELECT THE OPTION THAT REFLECTS HOW NERVOUS YOU ARE FEELING ABOUT YOUR UPCOMING KAMBO EXPERIENCES. | DEMO.16.",
    "PLEASE SELECT THE OPTION THAT REFLECTS HOW EXCITED YOU ARE FEELING ABOUT YOUR UPCOMING KAMBO EXPERIENCES. | DEMO.17."
]

# Export final standardized demographics dataset
df_cleaned = df_standardized.drop(columns=columns_to_drop, errors='ignore')
df_cleaned.to_csv("Kambo_demo_standardized.csv", index=False)

```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.preprocessing import StandardScaler

df = df_final.copy()

# Always keep these columns
id_cols = ['AnonID', 'Phase']

# Identify post-only variables (only appropriate for Phase 2)
post_only_cols = [col for col in df.columns if any(prefix in col for prefix in ['CEQ30_', 'MEQ30_', '5DASC_'])]

# Identify columns that should always be labeled DEMO_ regardless of phase
demo_only_cols = [
    col for col in df.columns
    if any(sub in col for sub in [
        'Session', 'Private_or_group', 'Vomit_', 'BM_', 'Fainting_',
        'Rapeh_', 'Sananga_', 'Age', 'Sex', 'Gender', 'Education',
        'Income', 'Height_in', 'Weight_lb', 'Lyme_DX', 'Post_Lyme_DX',
        'Psychedelic_Use', 'Cannabis_Use', 'Nervous', 'Excited', 'BMI', 'Days btwn', 'Autoimmu'
    ])
]

# Prepare phase-wise containers
phase_dfs = {}

for phase in [1, 2, 3]:
    df_phase = df[df['Phase'] == phase].copy()

    # Drop post-only data for Phases 1 and 3
    if phase != 2:
        df_phase = df_phase.drop(columns=post_only_cols, errors='ignore')

    # Tag columns with temporal prefix/suffix
    for col in df_phase.columns:
        if col in id_cols:
            continue
        if col in demo_only_cols:
            df_phase.rename(columns={col: f"DEMO_{col}"}, inplace=True)
        elif phase == 1:
            df_phase.rename(columns={col: f"PRE_{col}"}, inplace=True)
        elif phase == 2:
            df_phase.rename(columns={col: f"{col}_POST"}, inplace=True)
        elif phase == 3:
            df_phase.rename(columns={col: f"{col}_3MONTHSPOST"}, inplace=True)

    phase_dfs[phase] = df_phase



# Define imputation function
def impute_phase(df_phase):
    ids = df_phase[id_cols].reset_index(drop=True)
    numeric = df_phase.drop(columns=id_cols).select_dtypes(include=['float64', 'int64'])
    missing_before = numeric.isnull().sum()
    scaler = StandardScaler()
    scaled = pd.DataFrame(scaler.fit_transform(numeric), columns=numeric.columns)

    imputer = IterativeImputer(random_state=42, max_iter=10)
    imputed_array = imputer.fit_transform(scaled)

    # Only keep columns that actually have at least 1 non-missing value
    valid_cols = numeric.columns[numeric.notnull().any()].tolist()
    # After imputation, imputed_array.shape[1] == len(valid_cols)
    imputed = pd.DataFrame(imputed_array, columns=valid_cols)
    missing_after = imputed.isnull().sum()
    print("Missing per column before:", missing_before)
    print("Missing per column after:", missing_after)
    impute_fraction = (missing_before / len(numeric)).sort_values(ascending=False)
    print(impute_fraction)

    # Use only valid_cols for inverse transform
    unscaled = pd.DataFrame(
        scaler.inverse_transform(imputed.reindex(columns=scaled.columns, fill_value=0)),  # fill_value can be 0 or np.nan
        columns=scaled.columns
    )
    # If some columns were fully-missing, fill with NaN
    for col in scaled.columns:
        if col not in valid_cols:
            unscaled[col] = np.nan
    # Reorder columns (if needed)
    unscaled = unscaled[scaled.columns]

    return pd.concat([ids, unscaled], axis=1)

# Run and combine imputed phases
imputed_phases = [impute_phase(phase_dfs[p]) for p in [1, 2, 3]]
df_imputed_final = pd.concat(imputed_phases, axis=0).sort_values(by=['AnonID', 'Phase'])
columns_to_drop = ['PRE_Unnamed: 0', 'Unnamed: 0_POST', 'Unnamed: 0_3MONTHSPOST']
df_imputed_final.drop(columns=columns_to_drop, inplace=True)
df_imputed_final.head()
```
